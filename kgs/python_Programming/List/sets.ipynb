### Sets
-  **syntax of sets**
    - **note** : in sets the indexing and  slicing concept are not allowed. 
    - sets is not in hetroginios form 

- Unordered – Elements don't have a fixed position.
- No duplicates – Duplicate values are automatically removed.
- Mutable – You can add or remove elements.
- Uses { } curly braces.

```python
# ex:
s = {10,40,'apple',True}
print(s)
print(type(s))
s1 = {}
print(type(s1))
s2 = set ()
print(type(s2))
```

**common element Intersection**


```python
s1 = {1,2,3,4}
s2 = {3,4,5,6}
print(s1.union(s2))
print(s1.intersection(s2))
print(s1.difference(s2))
print(s2.difference(s1))
print(s1.symmetric_difference(s2))
s3 = {1,2,3,4}
s4 ={3,4,5,6}
s5 = {7,8,9,10}
print(s3.isdisjoint(s4))
print(s3.isdisjoint(s5))
print(s3.isdisjoint(s5))
```

```python
## super sets in subsets

s1 = {1,2,3,4}
s2 = {1,2}
print(s1.issuperset(s2))
print(s1.issubset(s2))
print(s2.issuperset(s1))
print(s2.issubset(s1))
```
