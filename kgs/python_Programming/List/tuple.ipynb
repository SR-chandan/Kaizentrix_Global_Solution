**TUPLE**

** TUPLE:**
- Heterogenous
- Duplicates are allowed
- immutable
- insertio

**Syntax:**
(ele 1 ,ele 2,_____ ele n)= {int,str,float,boolen}

```python
# ex:
t = (10,20,20.5,"apple",True)
print(t)
print(type(t))
print(len(t))
print(t[2])
t1 = (10)
print(type(t1))
print(t1)
t2 = (10,) # (, is used to convert the ingale int value to tuple.)
print(type(t2))
```

*Indexing and slicing in the tuple*

```python
t = (10,20,30,40,50,60)
print(t)
print(t[::1])
print(t[-2: :-1])
print(t[-4::+1])
print(t[-4:-1:-1])# no o/p
#print(t[-4:-2:0]) error as o/p
print(t[::-1])
print(t[2:-6:-1])
```

**Nested tuple**

```python
t =(10,20,30,(1.1,2.2,3.2,('a','b','c'),4.1),40)
print(t)
print(type(t))
print(len(t))
print(t[4])
print(t[3][0])
print(t[3][3][1])
```

**insertion and deletion in tuple**
- by using indexing and slicing 

```python
t = (10,20,30,40,50)
# insert 25 in[2:3]
t = t[: 2]+(25,)+ t[2:]
print(t)
t = t[:3]+t[4:]
print(t)
```

**zip function** in tuple 


```python
players = ['virat','dhoni','kl.rahul','rohith']
aeg = [35,41,31,39]
ipl_run = [9336,5439,5815,7329]
ipl_team = ['rcb','csk','lsg','mi']
info = list (zip(players,aeg,ipl_run,ipl_team))
print(info)
```

```python
#zip
state=['karnatak','tamilunadu','andraprdesh']
state_code=['KA','TN','AP']
capital=['bengaluru','chennai']
info=list(zip(state,state_code,capital))
print(info)
print("===========================================")

#zip_longest
from itertools import zip_longest
state=['karnatak','tamilunadu','andraprdesh']
state_code=['KA','TN','AP']
capital=['bengaluru','chennai']
info=list(zip_longest(state,state_code,capital))
print(info)
print("===========================================")

#fillvalue
from itertools import zip_longest
state=['karnatak','tamilunadu','andraprdesh']
state_code=['KA','TN','AP']
capital=['bengaluru','chennai']
info=list(zip_longest(state,state_code,capital,fillvalue="#"))#--in place # we can pass any value
print(info)
```
