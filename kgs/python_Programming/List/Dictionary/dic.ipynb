## Traversing in dict

```python
d = {'name':'rama','age':  23,'place':'banglore','mob': 916466}
for i in d:
    #print(i)----> only key 
    #print(d[i])#----> only values
    print(i,"  ",d[i]) #-----> keys and values 
```

##with the help of inbuilt ideas 

```python
d = {'name':'rama','age':  23,'place':'banglore','mob': 916466}
for i in d.values():
    print(i)
for i , j in d.items():
    print(i,"",j)
for i in d.keys():
    print(i)
```

## Nested dict 


```python
d = {'name':'rama','age':  23,'place':{'curr': 'banglore','per': 'hassan' },'mob':{'mob1': 916466,'mob2':9980}}
print(d)
print(d['name'])
print(d['place']['curr'])
d['place']['curr']='mysore'
print(d)
```

## copy in dictinory 
- shallow cpy
- deep cpy
 

```python
d = {'name':'rama','age':  23,'place':{'curr': 'banglore','per': 'hassan' },'mob':{'mob1': 916466,'mob2':9980}}
print(d)
print(d['name'])
print(d['place']['curr'])
d['place']['curr']='mysore'
print(d)

# copy

d1 = d 

print(d)
print(d1)
d2 = d.copy()
print(d2)
#d3 = copy.deepcopy(d)
#print(d3)
d['mob']['mob1']= 784569
print(d)
print(d1)
```

### ZIP  function in dictinory 

```python
emp_id = [101,102,103,104]
emp_name = ['rama','sita', 'ravana','abhi']
info = dict(zip(emp_id,emp_name))
print(info)


```

```python
emp_id = [101,102,103,104]
emp_name = ['rama','sita', 'ravana','abhi']
emp_salary = [2000,3000,40000,54000]
info = list(zip(emp_name,emp_salary))
print(info)
final_info = dict(zip(emp_id,info))
print(final_info)
```
