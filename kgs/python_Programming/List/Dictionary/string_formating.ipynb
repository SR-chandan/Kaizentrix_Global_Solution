## **Itrators and generotors** 

```python
# Itrators

s = ['chandna','abhi','chandra']
it = iter(s)
print(next(it))
print(next(it))
print(next(it))
a = ['chandna','abhi','chandra']
it = iter(a)
print(next(it))
print(next(it))
print(next(it))
```

```python
# Generotors

def number():
    yield 10
    yield 20 
    yield 30
g = number()
print(next(g))
print(next(g))
print(next(g))

print("==================================================")

# usinng for loop 

def number():
    yield 10
    yield 20 
    yield 30
for i in number():
    print(i)
```

**Iterable --> iterators --> next() --> one value at a time**
*** 
**Generators ---> func --> yield --> generators --> next() --> one value at a time**
