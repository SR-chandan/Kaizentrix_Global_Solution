```python
class Solution:
    def search(self, nums: list[int], target: int) -> bool:
        start, end = 0, len(nums) - 1

        while start <= end:
            mid = (start + end) // 2

            # Target found
            if nums[mid] == target:
                return True

            # Duplicates: cannot determine which half is sorted
            if nums[start] == nums[mid] == nums[end]:
                start += 1
                end -= 1
                continue

            # Left half is sorted
            if nums[start] <= nums[mid]:

                if nums[start] <= target < nums[mid]:
                    end = mid - 1
                else:
                    start = mid + 1

            # Right half is sorted
            else:

                if nums[mid] < target <= nums[end]:
                    start = mid + 1
                else:
                    end = mid - 1

        return False  
```

### Merge sort :

-**it majorly include 3 steps:**

    - diveide the array to its least form
        - LHS --> start = remains smae , end = prev mid
        - RHS--> start = (priv mid + 1),endd = remains same 
    - merge the array in sorted order inti a 3rd memory
    - update the sorted value's into the original memory 

**1. divide**,
    **merging**,
        **update original memory**

```python
def createIntArray():
    arr = []

    while True:
        try:
            n = int(input("Enter the val: "))
            arr.append(n)
        except Exception:
            return arr


# Merge two sorted parts
def mergeSortMerge(arr, start, mid, end):

    i = start
    j = mid + 1

    res = []

    # Compare both halves
    for k in range(end - start + 1):

        if i <= mid and j <= end:

            if arr[i] < arr[j]:
                res.append(arr[i])
                i += 1

            else:
                res.append(arr[j])
                j += 1

        else:

            if i <= mid:
                res.append(arr[i])
                i += 1

            elif j <= end:
                res.append(arr[j])
                j += 1

    # Update original array
    for k in range(len(res)):
        arr[start] = res[k]
        start += 1


# Divide
def mergeSortDivide(arr, start, end):

    if start >= end:
        return

    mid = (start + end) // 2

    # Left half
    mergeSortDivide(arr, start, mid)

    # Right half
    mergeSortDivide(arr, mid + 1, end)

    # Merge
    mergeSortMerge(arr, start, mid, end)


print("Enter ele's into the array:")

arr = createIntArray()

print("Created Array:", arr)

mergeSortDivide(arr, 0, len(arr) - 1)

print("Sorted Array:", arr)
```
