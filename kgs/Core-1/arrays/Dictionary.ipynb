## **Dictionary** 

- it is a data structure thet stores the data in key - value format 
- it is a iterable object 
- it is represented using --->{key:val}
- each pair of key - val is called as items
- it is also called collecrtion of items 
- len()--> it returns the count of num of ele's present in the itreable object.

## key :

- it is a unique ele's the represent each item individually.
- if we try to duplicate an existing key then the existing val's will get updated.
- muteble objects are not allowed to display all the key's are ==> dict.keys()
- it returns all the keys of the calling dictionary of types class --->dict.keys().

## values 

- it can be duplicated 
- it can be any datatype 
- the performed method to display all the key are ===> dict.values().
- it returns the all the keys of calling dictionary of type of class===>dict.values().
- to get all the items of the dictionary ===> dict.items().

```python
def createIntArray():
    l1 = []
    while True:
        try:
            n = int(input("Enter the num: "))
            l1.append(n)
        except Exception as e:
            return l1
def seggregation (arr):
    dict = {}
    for i in range(0,len(arr)):
        if arr [i] in dict:
            dict[arr[i]] = dict[arr[i]] + 1
        else:
            dict[arr[i]]= 1 
    #print(dict)
    dup , nondup , unique = [],[],[]
    #traversal through dictionary
    for key , val in dict.items():
        if val > 1 :
            dup.append(key)
        if  val == 1:
            unique.append(key)
        nondup.append(key)
    return dup, nondup,unique

print("enter the ele's into the array:")
arr = createIntArray()
print("created array: ",arr)

resDup,resNondup,resUnique = seggregation(arr)
print("duplicates:",resDup)
print("Non-dup:",resNondup)
print("Unique:",resUnique)
```
